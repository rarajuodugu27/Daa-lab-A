```python
import time

INF = float('inf')


def floyd_warshall(graph):
    """
    Floyd-Warshall Dynamic Programming algorithm for All-Pairs Shortest Paths.

    Time Complexity: O(V^3)
    Space Complexity: O(V^2)

    :param graph: 2D matrix where graph[i][j] is edge weight from i to j (INF if no edge, 0 if i == j)
    :return: (dist_matrix, next_matrix)
    """
    n = len(graph)
    # Initialize dist matrix and next node matrix for path reconstruction
    dist = [row[:] for row in graph]
    next_node = [[None] * n for _ in range(n)]

    for i in range(n):
        for j in range(n):
            if graph[i][j] != INF and i != j:
                next_node[i][j] = j

    # Dynamic Programming: iterate through all intermediate vertices k
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] != INF and dist[k][j] != INF:
                    if dist[i][k] + dist[k][j] < dist[i][j]:
                        dist[i][j] = dist[i][k] + dist[k][j]
                        next_node[i][j] = next_node[i][k]

    return dist, next_node


def reconstruct_path(u, v, next_node):
    """Reconstructs the shortest path from vertex u to vertex v."""
    if next_node[u][v] is None:
        return []
    path = [u]
    curr = u
    while curr != v:
        curr = next_node[curr][v]
        if curr is None:
            return []
        path.append(curr)
    return path


def print_distance_matrix(dist):
    """Prints formatted 2D distance matrix."""
    n = len(dist)
    print("\n========== ALL-PAIRS SHORTEST DISTANCE MATRIX ==========")
    header = f"{'From\\To':<8}" + "".join([f"{'Node ' + str(j):>10}" for j in range(n)])
    print(header)
    print("-" * len(header))
    for i in range(n):
        row = f"Node {i:<3}"
        for j in range(n):
            if dist[i][j] == INF:
                row += f"{'INF':>10}"
            else:
                row += f"{dist[i][j]:>10}"
        print(row)
    print("-" * len(header))


def main():
    print("==================================================")
    print("      FLOYD-WARSHALL ALL-PAIRS SHORTEST PATH      ")
    print("==================================================")

    # Default 4-vertex directed weighted graph
    # Node 0 -> Node 1 (3), Node 0 -> Node 3 (7)
    # Node 1 -> Node 0 (8), Node 1 -> Node 2 (2)
    # Node 2 -> Node 0 (5), Node 2 -> Node 3 (1)
    # Node 3 -> Node 0 (2)
    default_graph = [
        [0, 3, INF, 7],
        [8, 0, 2, INF],
        [5, INF, 0, 1],
        [2, INF, INF, 0]
    ]

    print("Default Graph Adjacency Weight Matrix:")
    print_distance_matrix(default_graph)

    user_choice = input("\nUse custom input? (y/N): ").strip().lower()
    if user_choice == 'y':
        try:
            n = int(input("Enter number of vertices: "))
            print(f"Enter {n}x{n} adjacency matrix row by row (use 'INF' for no edge):")
            graph = []
            for i in range(n):
                row_input = input(f"  Row {i}: ").split()
                row = [INF if val.upper() == 'INF' else int(val) for val in row_input]
                graph.append(row)
        except Exception as e:
            print(f"Invalid input ({e})! Falling back to default graph.")
            graph = default_graph
    else:
        graph = default_graph

    n = len(graph)
    print("\n--- Running Floyd-Warshall Dynamic Programming Solver ---")
    start_time = time.perf_counter()
    dist_matrix, next_matrix = floyd_warshall(graph)
    end_time = time.perf_counter()
    execution_time_ns = (end_time - start_time) * 1e9

    print_distance_matrix(dist_matrix)

    print("\n========== SAMPLE PATH RECONSTRUCTIONS ==========")
    print(f"{'Source':<8} -> {'Target':<8} | {'Cost':<6} | {'Path':<20}")
    print("-" * 48)
    for u in range(n):
        for v in range(n):
            if u != v:
                cost = "INF" if dist_matrix[u][v] == INF else str(dist_matrix[u][v])
                path = reconstruct_path(u, v, next_matrix)
                path_str = " -> ".join(map(str, path)) if path else "No Path"
                print(f"Node {u:<3} -> Node {v:<3} | {cost:<6} | {path_str:<20}")

    print("\n========== RESULTS SUMMARY ==========")
    print(f"Total Vertices (V)    : {n}")
    print(f"Execution Time        : {execution_time_ns:.2f} ns")
    print("======================================")


if __name__ == "__main__":
    main()
```
